```python
from array import *
arr = array("i", [])
x = int(input("enter size of array: "))
print("enter %d elements: " % x)

# এক লাইনে স্পেস দিয়ে ইনপুট নিয়ে লিস্ট বানানো হচ্ছে
elements = list(map(int, input().split()))

for i in range(x):
    arr.append(elements[i])

print(arr)
```
