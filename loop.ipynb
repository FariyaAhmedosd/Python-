##Matrix Create and 2 matrix sum


```python
r=int(input("Enter the number of rows:"))
c=int(input("Enter the number of columns:"))

#first Matrix (x) input neya
print("Enter elements for Matrix Y:")
x = []
for i in range(0, r):
    row = []
    for j in range(0, c):
        row.append(int(input(f"Enter the {i} * {j} element: ")))
    x.append(row)

# Dwitiyo Matrix (y) input neya
print("Enter elements for Matrix Y:")
y = []
for i in range(0, r):
    row = []
    for j in range(0, c):
        row.append(int(input(f"Enter the {i} * {j} element: ")))
    y.append(row)

# Matrix Duti Jog kora
matrix_sum = []
for i in range(0, r):
    row = []
    for j in range(0, c):
        row.append(x[i][j] + y[i][j])
    matrix_sum.append(row)

print("Sum of matrices:", matrix_sum)
```

#reverse of a number

```python
n=int(input("enter a number: "))
nr=0
while n%10!=0 :
  rem=n%10
  nr=nr*10+rem
  n=n//10
print(nr)
```

#Guessing number game

```python
import random

nump = random.randint(1000, 9999)
print("Generated number (for testing):", nump)  # Dorkar porle eta rakhte paro

# Prothomei user theke input nibe
n = int(input("Enter a 4 digit number (or enter 10 to quit): "))

while n != 10:
    # Original number guli ke preserve korar jonno temp variable use korchi
    temp_nump = nump
    temp_n = n
    cor = 0

    # 4 ta digit ekta ekta kore compare korar jonno loop
    while temp_nump > 0 and temp_n > 0:
        numc = temp_nump % 10
        nc = temp_n % 10

        if numc == nc:
            cor = cor + 1

        temp_nump = temp_nump // 10
        temp_n = temp_n // 10

    if cor == 4:
        print("Congrats! You guessed it right")
        break
    else:
        print(f"{cor} digits were guessed right")
        n = int(input("Enter a 4 digit number (or enter 10 to quit): "))
else:
    print("You quit the game")
```
