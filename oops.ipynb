##Encapsulation

```python
class car:

  def __init__(self, model, speed):
    self.model = model  # Public attribute
    self.__speed = (
        speed  # Encapsulated (private-style) attribute: বাইরে থেকে সরাসরি বদলানো যাবে না
    )

  def getmodel(self):
    print("The car model is:", self.model)

  # সুরক্ষিত __speed দেখার জন্য Getter মেথড
  def get_speed(self):
    print("Speed of the car is:", self.__speed)

  # নিয়ম মেনে __speed বদলানোর জন্য Setter মেথড
  def set_speed(self, new_speed):
    if new_speed > 0:
      self.__speed = new_speed
      print("Updated speed to:", self.__speed)
    else:
      print("Invalid speed! Speed must be greater than 0.")


class Sedan(car):  # child class

  def accelerate(self):
    print("137")

  def openTrunk(self):
    print("Trunk has been opened")


class SUV(car):

  def accelerate(self):
    print("127")


# Parent class-এর অবজেক্ট
obj1 = car("BMW", 180)
obj1.getmodel()
obj1.get_speed()

# বাইরে থেকে সরাসরি __speed বদলানোর চেষ্টা করলে কাজ করবে না (আসল মান সুরক্ষিত থাকবে)
obj1.__speed = 999
obj1.get_speed()  # এখনো 180 দেখাবে

# Setter মেথডের মাধ্যমে বৈধ মান দিয়ে স্পিড বদলানো
obj1.set_speed(200)
obj1.get_speed()  # এখন 200 দেখাবে

# অবৈধ (0 বা নেগেটিভ) মান দিলে আটকে দেবে
obj1.set_speed(-20)

print("---")

# Child class (Sedan)-এর অবজেক্টে Encapsulation-এর ব্যবহার
honda = Sedan("Honda Civic", 150)
honda.getmodel()
honda.get_speed()
honda.set_speed(165)
honda.accelerate()
honda.openTrunk()
```

#polymorphism

```python
class car:

  def __init__(self, model):
    self.model = model

  def accelerate(self):
    print("Car is accelerating...")


class Sedan(car):

  def accelerate(self):
    print(f"{self.model} (Sedan) accelerates smoothly at 137 km/h.")


class SUV(car):

  def accelerate(self):
    print(f"{self.model} (SUV) accelerates with high torque at 127 km/h.")


# আলাদা আলাদা অবজেক্ট
sedan_car = Sedan("Honda Civic")
suv_car = SUV("Toyota Fortuner")

# একই accelerate() মেথড কল করা হচ্ছে, কিন্তু ফল আলাদা আসছে
for vehicle in (sedan_car, suv_car):
  vehicle.accelerate()
```
